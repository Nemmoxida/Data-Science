# Insight
1. rata rata skor kinerja karyawan adalah 68.36 dengan std deviasi 12.37. artinya range rata ratanya adalah 56 - 70
2. sales memiliki sebaran rata rata skor kinerja paling stabil di antara semua departemen
3. departemen finance, sales, dan operasional memiliki median tertinggi dalam kategori lama bekerja, diangka 8 tahun

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Dataset_Kinerja_Karyawan.csv")
df.info()
```

```python
df.head()
```

```python
print("Statistik deskriptif kinerja karyawan")
df['Skor_Kinerja'].describe()
```

```python
skor_kinerja = df['Skor_Kinerja']
```

```python
from numpy import sqrt
modus = skor_kinerja.mode()[0]
varians = skor_kinerja.var()
range = skor_kinerja.max() - skor_kinerja.min()

print("Skor Kinerja Karyawan")
print("Modus:", modus)
print("Varians:", f"{varians:.2f}")
print("Standar Deviasi:", f"{sqrt(varians):.2f}")
print("Range:", range)

```

```python
plt.figure(figsize=(8, 6))
sns.histplot(skor_kinerja, bins=20, kde=True, color="skyblue", edgecolor="black")
plt.axvline(skor_kinerja.mean(), color='red', linestyle='dashed', linewidth=2, label=f'Mean: {skor_kinerja.mean():.2f}')
plt.axvline(skor_kinerja.median(), color='green', linestyle='dashed', linewidth=2, label=f'Median: {skor_kinerja.median():.2f}')
plt.title("Distribusi Skor Kinerja Karyawan", fontsize=15, pad=10)
plt.xlabel("Skor Kinerja", fontsize=12)
plt.ylabel("Frekuensi", fontsize=12)
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='Departemen', y="Skor_Kinerja", data=df, palette="Set2")
plt.title("Sebaran Skor Kinerja Per Departemen", fontsize=14, pad=15)
plt.xlabel("Departemen", fontsize=12)
plt.ylabel("Skor Kinerja", fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='Departemen', y="Lama_Bekerja_Tahun", data=df, palette="viridis")
plt.title("Rata rata Bekerja Karyawan per Departemen", fontsize=14, pad=15)
plt.xlabel("Departemen", fontsize=12)
plt.ylabel("Lama Bekerja (Tahun)", fontsize=12)
plt.show()
```
